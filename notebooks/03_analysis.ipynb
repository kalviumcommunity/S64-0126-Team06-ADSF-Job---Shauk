# 03 — Analysis

Compute skill frequencies, placement rates, and trends.

```python
import pandas as pd

# Load the cleaned data
df_clean = pd.read_csv('../data/processed/cleaned_job_postings.csv')

# Convert skills_list back to list (since CSV saves as string)
df_clean['skills_list'] = df_clean['skills_required'].str.split(',')

# Skill frequencies
all_skills = df_clean['skills_list'].explode()
skill_freq = all_skills.value_counts()
print("Skill Frequencies:")
print(skill_freq)

# Average salary by sector (placement rates proxy)
avg_salary_by_sector = df_clean.groupby('sector')['salary_lpa'].mean()
print("\nAverage Salary by Sector:")
print(avg_salary_by_sector)

# Trends: Salary over time (sort by date)
df_clean['date_posted'] = pd.to_datetime(df_clean['date_posted'])
df_sorted = df_clean.sort_values('date_posted')
print("\nSalary Trends (sorted by date):")
print(df_sorted[['date_posted', 'salary_lpa', 'job_title']])

# Additional insights: Top companies, etc.
top_sectors = df_clean['sector'].value_counts()
print("\nSector Distribution:")
print(top_sectors)

# Save analysis results (optional)
analysis_results = {
    'skill_frequencies': skill_freq.to_dict(),
    'avg_salary_by_sector': avg_salary_by_sector.to_dict(),
    'sector_distribution': top_sectors.to_dict()
}
import json
with open('../outputs/analysis_results.json', 'w') as f:
    json.dump(analysis_results, f, indent=4)

print("\nAnalysis completed. Results saved to outputs/analysis_results.json")
```
