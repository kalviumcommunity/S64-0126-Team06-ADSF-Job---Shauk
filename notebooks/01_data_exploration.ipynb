# 01 — Data Exploration

Load raw data and inspect shape, dtypes, and null counts.

```python
import pandas as pd
df = pd.read_csv('../data/raw/sample_job_postings.csv')
print(df.head())

# Inspect the shape
print(f"\nDataset shape: {df.shape}")

# Inspect data types
print(f"\nData types:\n{df.dtypes}")

# Check for null values
print(f"\nNull counts:\n{df.isnull().sum()}")
```
