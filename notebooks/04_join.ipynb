# Stage 4 — Join

Build the skill demand-supply table and sector-skill matrix.

```python
import sys
from pathlib import Path
import pandas as pd
import ast
sys.path.insert(0, str(Path(".").resolve()))

from src.utils import build_skill_demand_supply_table, build_sector_skill_matrix
```

```python
df_jobs = pd.read_csv("data/interim/jobs_canonicalized.csv")
df_placements = pd.read_csv("data/interim/placements_canonicalized.csv")
for col in ["canonical_skills"]:
    df_jobs[col] = df_jobs[col].apply(ast.literal_eval)
    df_placements[col] = df_placements[col].apply(ast.literal_eval)
```

## Demand-supply table

```python
skill_ds = build_skill_demand_supply_table(df_jobs, df_placements)
print(f"Skills in output: {len(skill_ds)}")
skill_ds.sort_values("lift", ascending=False).head(10)
```

## Sector-skill matrix

```python
sector_matrix = build_sector_skill_matrix(df_jobs, df_placements)
print(sector_matrix.shape)
sector_matrix
```

## Save processed outputs

```python
skill_ds.to_csv("data/processed/skill_demand_supply.csv", index=False)
sector_matrix.to_csv("data/processed/sector_skill_matrix.csv")
print("Saved.")
```
