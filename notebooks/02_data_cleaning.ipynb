# 02 — Data Cleaning

Clean, deduplicate, and standardise the raw data.

```python
import pandas as pd

# Load the raw data
df = pd.read_csv('../data/raw/sample_job_postings.csv')

# Check for duplicates
print(f"Number of duplicate rows: {df.duplicated().sum()}")

# Remove duplicates if any
df_clean = df.drop_duplicates()

# Standardize column names (lowercase, replace spaces with underscores)
df_clean.columns = df_clean.columns.str.lower().str.replace(' ', '_')

# Handle missing values (example: fill with 'Unknown' for categorical, mean for numeric)
df_clean['company'] = df_clean['company'].fillna('Unknown')
df_clean['sector'] = df_clean['sector'].fillna('Unknown')
df_clean['skills_required'] = df_clean['skills_required'].fillna('Not Specified')
# For salary, if missing, perhaps drop or impute
df_clean = df_clean.dropna(subset=['salary_lpa'])  # Drop rows with missing salary

# Convert date_posted to datetime
df_clean['date_posted'] = pd.to_datetime(df_clean['date_posted'])

# Standardize skills_required (split into list for easier analysis later)
df_clean['skills_list'] = df_clean['skills_required'].str.split(',')

print("Data cleaning completed.")
print(f"Original shape: {df.shape}")
print(f"Cleaned shape: {df_clean.shape}")

# Save cleaned data
df_clean.to_csv('../data/processed/cleaned_job_postings.csv', index=False)
```
